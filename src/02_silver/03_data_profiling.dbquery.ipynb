```sql
%sql
CREATE OR REPLACE TEMP VIEW data_quality_report AS

-- ============================================================
-- AISLES
-- ============================================================

SELECT
    'aisles' AS table_name,
    'aisle_id' AS column_name,
    'BIGINT' AS data_type,
    COUNT(*) AS total_rows,
    COUNT(*) - COUNT(aisle_id) AS null_count,
    ROUND(
        100.0 * (COUNT(*) - COUNT(aisle_id))
        / COUNT(*), 2
    ) AS null_percentage,
    COUNT(DISTINCT aisle_id) AS distinct_count,
    ROUND(
        100.0 * COUNT(DISTINCT aisle_id)
        / COUNT(*), 2
    ) AS distinct_percentage,
    CAST(MIN(aisle_id) AS STRING) AS min_value,
    CAST(MAX(aisle_id) AS STRING) AS max_value
FROM instacart.aisles

UNION ALL

SELECT
    'aisles',
    'aisle',
    'STRING',
    COUNT(*),
    COUNT(*) - COUNT(aisle),
    ROUND(
        100.0 * (COUNT(*) - COUNT(aisle))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT aisle),
    ROUND(
        100.0 * COUNT(DISTINCT aisle)
        / COUNT(*), 2
    ),
    MIN(aisle),
    MAX(aisle)
FROM instacart.aisles


-- ============================================================
-- DEPARTMENTS
-- ============================================================

UNION ALL

SELECT
    'departments',
    'department_id',
    'BIGINT',
    COUNT(*),
    COUNT(*) - COUNT(department_id),
    ROUND(
        100.0 * (COUNT(*) - COUNT(department_id))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT department_id),
    ROUND(
        100.0 * COUNT(DISTINCT department_id)
        / COUNT(*), 2
    ),
    CAST(MIN(department_id) AS STRING),
    CAST(MAX(department_id) AS STRING)
FROM instacart.departments

UNION ALL

SELECT
    'departments',
    'department',
    'STRING',
    COUNT(*),
    COUNT(*) - COUNT(department),
    ROUND(
        100.0 * (COUNT(*) - COUNT(department))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT department),
    ROUND(
        100.0 * COUNT(DISTINCT department)
        / COUNT(*), 2
    ),
    MIN(department),
    MAX(department)
FROM instacart.departments


-- ============================================================
-- PRODUCTS
-- ============================================================

UNION ALL

SELECT
    'products',
    'product_id',
    'BIGINT',
    COUNT(*),
    COUNT(*) - COUNT(product_id),
    ROUND(
        100.0 * (COUNT(*) - COUNT(product_id))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT product_id),
    ROUND(
        100.0 * COUNT(DISTINCT product_id)
        / COUNT(*), 2
    ),
    CAST(MIN(product_id) AS STRING),
    CAST(MAX(product_id) AS STRING)
FROM instacart.products

UNION ALL

SELECT
    'products',
    'product_name',
    'STRING',
    COUNT(*),
    COUNT(*) - COUNT(product_name),
    ROUND(
        100.0 * (COUNT(*) - COUNT(product_name))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT product_name),
    ROUND(
        100.0 * COUNT(DISTINCT product_name)
        / COUNT(*), 2
    ),
    MIN(product_name),
    MAX(product_name)
FROM instacart.products

UNION ALL

SELECT
    'products',
    'aisle_id',
    'BIGINT',
    COUNT(*),
    COUNT(*) - COUNT(aisle_id),
    ROUND(
        100.0 * (COUNT(*) - COUNT(aisle_id))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT aisle_id),
    ROUND(
        100.0 * COUNT(DISTINCT aisle_id)
        / COUNT(*), 2
    ),
    CAST(MIN(aisle_id) AS STRING),
    CAST(MAX(aisle_id) AS STRING)
FROM instacart.products

UNION ALL

SELECT
    'products',
    'department_id',
    'BIGINT',
    COUNT(*),
    COUNT(*) - COUNT(department_id),
    ROUND(
        100.0 * (COUNT(*) - COUNT(department_id))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT department_id),
    ROUND(
        100.0 * COUNT(DISTINCT department_id)
        / COUNT(*), 2
    ),
    CAST(MIN(department_id) AS STRING),
    CAST(MAX(department_id) AS STRING)
FROM instacart.products


-- ============================================================
-- ORDERS
-- ============================================================

UNION ALL

SELECT
    'orders',
    'order_id',
    'BIGINT',
    COUNT(*),
    COUNT(*) - COUNT(order_id),
    ROUND(
        100.0 * (COUNT(*) - COUNT(order_id))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT order_id),
    ROUND(
        100.0 * COUNT(DISTINCT order_id)
        / COUNT(*), 2
    ),
    CAST(MIN(order_id) AS STRING),
    CAST(MAX(order_id) AS STRING)
FROM instacart.orders

UNION ALL

SELECT
    'orders',
    'user_id',
    'BIGINT',
    COUNT(*),
    COUNT(*) - COUNT(user_id),
    ROUND(
        100.0 * (COUNT(*) - COUNT(user_id))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT user_id),
    ROUND(
        100.0 * COUNT(DISTINCT user_id)
        / COUNT(*), 2
    ),
    CAST(MIN(user_id) AS STRING),
    CAST(MAX(user_id) AS STRING)
FROM instacart.orders

UNION ALL

SELECT
    'orders',
    'order_number',
    'BIGINT',
    COUNT(*),
    COUNT(*) - COUNT(order_number),
    ROUND(
        100.0 * (COUNT(*) - COUNT(order_number))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT order_number),
    ROUND(
        100.0 * COUNT(DISTINCT order_number)
        / COUNT(*), 2
    ),
    CAST(MIN(order_number) AS STRING),
    CAST(MAX(order_number) AS STRING)
FROM instacart.orders

UNION ALL

SELECT
    'orders',
    'order_dow',
    'BIGINT',
    COUNT(*),
    COUNT(*) - COUNT(order_dow),
    ROUND(
        100.0 * (COUNT(*) - COUNT(order_dow))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT order_dow),
    ROUND(
        100.0 * COUNT(DISTINCT order_dow)
        / COUNT(*), 2
    ),
    CAST(MIN(order_dow) AS STRING),
    CAST(MAX(order_dow) AS STRING)
FROM instacart.orders

UNION ALL

SELECT
    'orders',
    'order_hour_of_day',
    'BIGINT',
    COUNT(*),
    COUNT(*) - COUNT(order_hour_of_day),
    ROUND(
        100.0 * (COUNT(*) - COUNT(order_hour_of_day))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT order_hour_of_day),
    ROUND(
        100.0 * COUNT(DISTINCT order_hour_of_day)
        / COUNT(*), 2
    ),
    CAST(MIN(order_hour_of_day) AS STRING),
    CAST(MAX(order_hour_of_day) AS STRING)
FROM instacart.orders

UNION ALL

SELECT
    'orders',
    'days_since_prior_order',
    'DOUBLE',
    COUNT(*),
    COUNT(*) - COUNT(days_since_prior_order),
    ROUND(
        100.0 * (COUNT(*) - COUNT(days_since_prior_order))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT days_since_prior_order),
    ROUND(
        100.0 * COUNT(DISTINCT days_since_prior_order)
        / COUNT(*), 2
    ),
    CAST(MIN(days_since_prior_order) AS STRING),
    CAST(MAX(days_since_prior_order) AS STRING)
FROM instacart.orders


-- ============================================================
-- ORDER_PRODUCTS
-- ============================================================

UNION ALL

SELECT
    'order_products',
    'order_id',
    'BIGINT',
    COUNT(*),
    COUNT(*) - COUNT(order_id),
    ROUND(
        100.0 * (COUNT(*) - COUNT(order_id))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT order_id),
    ROUND(
        100.0 * COUNT(DISTINCT order_id)
        / COUNT(*), 2
    ),
    CAST(MIN(order_id) AS STRING),
    CAST(MAX(order_id) AS STRING)
FROM instacart.order_products

UNION ALL

SELECT
    'order_products',
    'product_id',
    'BIGINT',
    COUNT(*),
    COUNT(*) - COUNT(product_id),
    ROUND(
        100.0 * (COUNT(*) - COUNT(product_id))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT product_id),
    ROUND(
        100.0 * COUNT(DISTINCT product_id)
        / COUNT(*), 2
    ),
    CAST(MIN(product_id) AS STRING),
    CAST(MAX(product_id) AS STRING)
FROM instacart.order_products

UNION ALL

SELECT
    'order_products',
    'add_to_cart_order',
    'BIGINT',
    COUNT(*),
    COUNT(*) - COUNT(add_to_cart_order),
    ROUND(
        100.0 * (COUNT(*) - COUNT(add_to_cart_order))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT add_to_cart_order),
    ROUND(
        100.0 * COUNT(DISTINCT add_to_cart_order)
        / COUNT(*), 2
    ),
    CAST(MIN(add_to_cart_order) AS STRING),
    CAST(MAX(add_to_cart_order) AS STRING)
FROM instacart.order_products

UNION ALL

SELECT
    'order_products',
    'reordered',
    'BIGINT',
    COUNT(*),
    COUNT(*) - COUNT(reordered),
    ROUND(
        100.0 * (COUNT(*) - COUNT(reordered))
        / COUNT(*), 2
    ),
    COUNT(DISTINCT reordered),
    ROUND(
        100.0 * COUNT(DISTINCT reordered)
        / COUNT(*), 2
    ),
    CAST(MIN(reordered) AS STRING),
    CAST(MAX(reordered) AS STRING)
FROM instacart.order_products
;


/* ============================================================
   FINAL DATA QUALITY REPORT
   ============================================================ */

SELECT
    table_name,
    column_name,
    data_type,
    total_rows,
    null_count,
    null_percentage,
    distinct_count,
    distinct_percentage,
    min_value,
    max_value

FROM data_quality_report

ORDER BY
    table_name,
    column_name;
```
