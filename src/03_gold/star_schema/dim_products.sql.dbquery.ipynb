```sql
-- Dimension: Products (denormalized with aisle and department)
CREATE OR REPLACE TABLE instacart_gold.products AS
SELECT
  p.product_id,
  p.product_name,
  p.aisle_id,
  COALESCE(a.aisle, 'Unknown') AS aisle,
  p.department_id,
  COALESCE(d.department, 'Unknown') AS department
FROM instacart_silver.products p
LEFT JOIN instacart_silver.aisles a ON p.aisle_id = a.aisle_id
LEFT JOIN instacart_silver.departments d ON p.department_id = d.department_id;
```
