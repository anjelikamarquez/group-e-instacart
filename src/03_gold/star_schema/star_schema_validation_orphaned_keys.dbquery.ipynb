```sql
-- Validate referential integrity: check for orphaned foreign keys
SELECT
  'order_id' AS foreign_key,
  COUNT(DISTINCT op.order_id) AS fact_distinct_values,
  COUNT(DISTINCT o.order_id) AS dim_distinct_values,
  COUNT(DISTINCT op.order_id) - COUNT(DISTINCT o.order_id) AS orphaned_records
FROM instacart_gold.order_products op
LEFT JOIN instacart_gold.orders o ON op.order_id = o.order_id

UNION ALL

SELECT
  'product_id',
  COUNT(DISTINCT op.product_id),
  COUNT(DISTINCT p.product_id),
  COUNT(DISTINCT op.product_id) - COUNT(DISTINCT p.product_id)
FROM instacart_gold.order_products op
LEFT JOIN instacart_gold.products p ON op.product_id = p.product_id;
```
