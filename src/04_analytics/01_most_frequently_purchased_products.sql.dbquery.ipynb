```sql
-- Q1: Which products and departments are purchased most frequently?

-- Top 20 Products by Order Frequency
SELECT
  p.product_name,
  p.department,
  p.aisle,
  COUNT(DISTINCT op.order_id) AS times_ordered,
  COUNT(*) AS total_units,
  ROUND(SUM(op.reordered) * 100.0 / COUNT(*), 2) AS reorder_rate_pct
FROM instacart_gold.order_products op
INNER JOIN instacart_gold.products p ON op.product_id = p.product_id
GROUP BY p.product_name, p.department, p.aisle
ORDER BY times_ordered DESC
LIMIT 20;
```
