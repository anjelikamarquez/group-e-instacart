```sql
-- Q3: Which products have the highest reorder behavior?

-- Top 10 Products by Reorder Rate (minimum 100 orders)
SELECT
  p.product_name,
  p.department,
  p.aisle,
  COUNT(*) AS total_units_sold,
  SUM(op.reordered) AS times_reordered,
  ROUND(SUM(op.reordered) * 100.0 / COUNT(*), 2) AS reorder_rate_pct,
  COUNT(DISTINCT op.order_id) AS unique_orders
FROM instacart_gold.order_products op
INNER JOIN instacart_gold.products p ON op.product_id = p.product_id
GROUP BY p.product_name, p.department, p.aisle
HAVING COUNT(DISTINCT op.order_id) >= 100  -- Filter for statistical significance
ORDER BY reorder_rate_pct DESC, unique_orders DESC
LIMIT 10;
```
