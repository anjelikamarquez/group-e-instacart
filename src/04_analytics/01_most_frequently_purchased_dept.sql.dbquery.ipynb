```sql
-- Top Departments by Order Frequency
SELECT
  p.department,
  COUNT(DISTINCT op.order_id) AS orders_with_dept,
  COUNT(*) AS total_products_sold,
  COUNT(DISTINCT op.product_id) AS unique_products,
  ROUND(AVG(op.add_to_cart_order), 2) AS avg_cart_position,
  ROUND(SUM(op.reordered) * 100.0 / COUNT(*), 2) AS reorder_rate_pct
FROM instacart_gold.order_products op
INNER JOIN instacart_gold.products p ON op.product_id = p.product_id
GROUP BY p.department
ORDER BY orders_with_dept DESC; 
```
