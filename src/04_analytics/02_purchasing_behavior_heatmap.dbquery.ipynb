```sql
-- Day of Week × Hour of Day (for heatmap visualization)
SELECT
  CASE o.order_dow
    WHEN 0 THEN 'Sunday'
    WHEN 1 THEN 'Monday'
    WHEN 2 THEN 'Tuesday'
    WHEN 3 THEN 'Wednesday' 
    WHEN 4 THEN 'Thursday'
    WHEN 5 THEN 'Friday'
    WHEN 6 THEN 'Saturday'
  END AS day_of_week,
  o.order_hour_of_day,
  COUNT(DISTINCT o.order_id) AS total_orders,
  ROUND(COUNT(op.product_id) * 1.0 / COUNT(DISTINCT o.order_id), 2) AS avg_basket_size
FROM instacart_gold.orders o
INNER JOIN instacart_gold.order_products op ON o.order_id = op.order_id
GROUP BY o.order_dow, o.order_hour_of_day
ORDER BY o.order_dow, o.order_hour_of_day;
```
